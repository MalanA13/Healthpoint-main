# 01 - Gathering Data

Notebook ini mendokumentasikan proses gathering dataset appointment pasien untuk HealPoint. Dataset dipilih karena sesuai dengan problem bisnis proyek: mengurangi ketidakhadiran pasien dan membantu sistem scheduling berbasis data.

## Sumber Data

- File sumber: `../../KaggleV2-May-2016.csv`
- Jumlah baris awal: **110,527**
- Jumlah kolom awal: **14**
- Target analisis: kolom `No-show`, yaitu apakah pasien tidak hadir pada appointment.

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('../../KaggleV2-May-2016.csv')
df = pd.read_csv(DATA_PATH, dtype={'PatientId': str})
df.head()
```

## Deskripsi Kolom Awal

Dataset memiliki identifier pasien dan appointment, informasi demografis dasar, tanggal pembuatan jadwal, tanggal appointment, kondisi kesehatan, status SMS reminder, dan target `No-show`.

```python
df[['Gender', 'Age', 'Neighbourhood', 'ScheduledDay', 'AppointmentDay', 'SMS_received', 'No-show']].sample(10, random_state=42)
```
