# 04 - Feature Engineering

Notebook ini menjelaskan fitur turunan yang dipakai untuk model AI dan memastikan tidak terjadi data leakage.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/appointments_clean.csv')
df.head()
```

## Fitur Turunan

- `waiting_days`: selisih hari antara jadwal dibuat dan hari appointment.
- `age_group`: segmentasi usia pasien.
- `has_chronic_condition`: gabungan indikator hypertension atau diabetes.
- `scheduled_hour`: jam saat jadwal dibuat.
- `appointment_weekday`: hari appointment.

```python
engineered_cols = ['waiting_days', 'age_group', 'has_chronic_condition', 'scheduled_hour', 'appointment_weekday']
df[engineered_cols].head(10)
```

```python
df['age_group'].value_counts().plot(kind='bar', title='Age Group Distribution')
plt.show()
```

```python
df['scheduled_hour'].value_counts().sort_index().plot(kind='bar', title='Scheduled Hour Distribution')
plt.show()
```

```python
df['has_chronic_condition'].value_counts(normalize=True).mul(100).round(2)
```

## Data Leakage Check

Kolom yang tidak digunakan sebagai fitur model:

- `is_no_show`: target, tidak boleh masuk fitur.
- `appointment_id`: identifier.
- `patient_id`: identifier.
- `scheduled_day` dan `appointment_day`: datetime mentah diganti fitur turunan agar model tidak bergantung pada identifier waktu spesifik.

```python
features = ['gender', 'age', 'age_group', 'neighbourhood', 'scheduled_hour', 'appointment_weekday', 'appointment_month', 'waiting_days', 'scholarship', 'hypertension', 'diabetes', 'alcoholism', 'handicap', 'sms_received', 'has_chronic_condition']
target = 'is_no_show'
model_ready = df[features + [target]]
model_ready.to_csv('../data/processed/model_ready_features.csv', index=False)
model_ready.to_csv('../../ai-engineer/model/saved_model/appointments_model_ready.csv', index=False)
model_ready.head()
```
